Series

```python
import pandas as pd

#Creating series
sales=pd.Series([1,2,3,4,5])
print(sales)


##indexing in series
print(sales[2])

##Adding custom labels ans accessing the elements
sales=pd.Series([1,2,3,4],index=["a","b","c","d"])
print(sales["c"])

#Slicing
s = pd.Series([10, 20, 30, 40, 50])

print(s[1:4]) ##last index is not taken into account in slicing

##Methods and attributes in pandas

s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s)
print(s.index)    # Index(['a', 'b', 'c', 'd'])
print(s.values)   # [10 20 30 40]
print(s.dtype)    # int64 (typically)
print(s.shape)    # (4,)
print(s.size)     # 4
print(s.ndim)     # 1

print(s.head(2))   # First 2 elements
print(s.tail(2))   # Last 2 elements
print(s.describe())
print(s.to_list()) # [10, 20, 30, 40]

print(s.loc['a'])       # 10
print(s.iloc[2])        # 30
print(s.isin([10, 30])) # True, False, True, False
print(s.between(20, 40))
```

Dataframes

```python
import pandas as pd

##Creating a dataframe
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Sleep_Hours': [8, 7, 7, 6, 6],
    'Attendance': [60, 70, 80, 90, 95],
    'Score': [40, 50, 65, 80, 95]
}

df = pd.DataFrame(data)

print(df)

#Attributes and methods related to dataframe
df.shape
df.columns
df.dtypes

## Accessing data in pandas using labels


# loc
print(df.loc[2])                                   # row with label 2
print(df.loc[2, 'Score'])                          # single value: row 2, Score -> 65
print(df.loc[1:3])                                 # rows 1, 2, 3 (end label INCLUDED)
print(df.loc[:, 'Score'])                          # whole Score column
print(df.loc[:, ['Study_Hours', 'Score']])         # two columns by name
print(df.loc[1:3, 'Study_Hours':'Attendance'])     # rows 1-3, columns Study_Hours to Attendance (end INCLUDED)
print(df.loc[df['Score'] > 60])                    # boolean filter: students with Score above 60
print(df.loc[df['Score'] > 60, ['Study_Hours', 'Score']])  # filter rows and pick two columns

# iloc
print(df.iloc[2])                                  # third row (position 2)
print(df.iloc[2, 3])                               # row position 2, column position 3 -> 65
print(df.iloc[1:3])                                # rows at positions 1 and 2 (end EXCLUDED)
print(df.iloc[:, 3])                               # column at position 3 (Score)
print(df.iloc[:, [0, 3]])                          # columns at positions 0 and 3
print(df.iloc[0:3, 0:2])                           # first 3 rows, first 2 columns
print(df.iloc[-1])                                 # last row

                
```
