```python
import pandas as pd
data = {
        'Ad_Budget': [2, 4, 6, 8, 10],
        'Discount': [5, 8, 10, 15, 20],
        'Customers': [120, 150, 190, 230, 260],
        'Sales': [35, 48, 62, 81, 97]
    }
df = pd.DataFrame(data)
df
```

PART 1: Selecting by Label (.loc)

```python
#Write code to get the Customers value of the row labeled 3.
print(df.loc[3,'Customers'])

#Select rows 0 to 2 with only the Discount and Sales columns.
print(df.loc[:2,["Discount","Sales"]])

#How many rows does df.loc[1:3, 'Sales'] return, and why?
print(df.loc[1:3,'Sales']) #Selects the row from 1 to 3 of sales column only

#Why does df.loc[1:3] return 3 rows while df.iloc[1:3] returns 2?
print(df.loc[1:3]) #Includes the ending element
print(df.iloc[1:3]) #Excludes the ending element

```

PART 2: Selecting by Position (.iloc)

```python
#Get the Ad_Budget value of the last row using .iloc.
print(df.iloc[4,0])

#Select the first 3 rows and first 2 columns.
print(df.iloc[:3,:1])

#Create X (all columns except Sales) and y (only Sales) using .iloc.
X=df.iloc[:,:3]
Y=df.iloc[:,-1]
print("All column except sales: \n",X)
print("Only sales column\n",Y)

# What does df.iloc[:, -1] return?
print(df.iloc[:,-1]) #Returns only the sales column

```

PART 3: Filtering Rows

```python
#Show all rows with Ad_Budget >= 6
print(df[df["Ad_Budget"]>=6])

#Show rows with Discount > 10 AND Sales > 60.
print(df[(df['Discount']>10) & (df['Sales']>60)])

#Show rows with Sales < 50 OR Customers > 230 (use |).
print(df[(df['Sales']<50) | (df['Customers']>230)])

# #Show only Ad_Budget and Customers for rows where Sales is between 50 and 80 (inclusive).
# result=df.loc[df[(df['Sales']) >50 & (df['Sales']<80)],['Ad_Budget','Customers']]
# print(result)


# Show only Ad_Budget and Customers for rows where Sales is between 50 and 80 (inclusive).
result = df.loc[
    (df['Sales'] >= 50) & (df['Sales'] <= 80)
    ,['Ad_Budget', 'Customers']
    ]     
print(result)
     
#Memorize this pattern
# result = df.loc[
#     (condition1) & (condition2),
#     ['column1', 'column2']
# ]


# This code gives an error. Why, and how do you fix it?
     # df[df['Sales'] > 50 & df['Customers'] > 200]
     
fixed=df[(df['Sales'] > 50) & (df['Customers'] > 200)]
print(fixed)


```
