Adding and removing the data 


```python
import numpy as np 

## Adding elements to an 1D array
arr = np.array([10, 20, 30, 40])
new_arr = np.append(arr, 50)

print(new_arr)

## Adding elements to an 2D array

arr = np.array([
    [10, 20],
    [30, 40]
])
print("Original array : ",arr)
new_arrRow = np.append(arr, [[50, 60]], axis=0) ## Add a new row
new_arrColumn = np.append(arr, [[50], [60]], axis=1) ## Add a new column

print("Added a new row: ",new_arrRow)
print("Added a new column: ",new_arrColumn)


##Removing elements from an array

arr = np.array([10, 20, 30, 40, 50])

new_arr = np.delete(arr, 2)

print(new_arr)

##Removing from an 2D array is same as 1D but just use axis
```

Combining arrays

```python
import numpy as np

## Combining arrays in 1D 
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])

result = np.concatenate((arr1, arr2))

print(result)

##Combining in 2D arrays

arr1 = np.array([
    [10, 20],
    [30, 40]
])

arr2 = np.array([
    [50, 60],
    [70, 80]
])

resultVerti = np.concatenate((arr1, arr2), axis=0) ##Combine vertically
resultHoriz = np.concatenate((arr1, arr2), axis=1) ##Combine horizaontally

print("Combined vetically: ",resultVerti)
print("Combined horizontally: ",resultHoriz)


##Combining arrays using vstack and hstack

resultV = np.vstack((arr1, arr2))

print("Combined vertically using vstack: ",resultV)

resultH = np.hstack((arr1, arr2))

print("Combined Horizontally using hstack: ",resultH)
```
