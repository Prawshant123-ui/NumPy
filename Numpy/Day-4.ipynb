## Practice with real world problem

```python
import numpy as np

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000],  # Chai Point
    [6, 95000,  125000, 150000, 175000],  # Momo House
    [7, 220000, 250000, 280000, 320000],  # Thakali Kitchen
    [8, 110000, 135000, 170000, 200000],  # Sushi Corner
    [9, 175000, 190000, 215000, 245000],  # Cafe Mocha
    [10, 130000, 155000, 185000, 210000]  # Tandoori Palace
])
```

Indexing and Slicing Questions 

```python
## Get the first restaurant's complete record
print("Complete record of 1st restaurant: ", sales_data[0])

##Get the last restaurant's complete record.
print("Complete record of last restaurant: ", sales_data[-1])

##Get sales for all restaurants in 2022.
print("Sales of all restaurants in 2022 : ",sales_data[:,2])

##Get sales for all restaurants in 2024.
print("Sales of all restaurant in 2024 : ",sales_data[:,-1])

##Get records of restaurants 1–5
print("Record of restaurants 1 to 5 : ",sales_data[0:5])

##Get only the 2022 and 2023 sales columns.
print("sales column of only 2022 and 2023 : ",sales_data[:,2:-1])

##Get the 2024 sales of restaurants 3, 5, and 7.
print("2024 Sales of restaurants 3,5 and 7: ",sales_data[2:7:2,-1])
```

Calculation questions 

```python
##Find the average sales in 2021.
all_sales=sales_data[:,1]
avg_sales=np.average(all_sales)
print("Average sales in 2021: ",avg_sales)

##Find the highest sales in 2024.
all_sales=sales_data[:,-1]
max_sales=np.max(all_sales)
print("Maximum sales in 2024: ",max_sales)

##Find the lowest sales in 2022.
all_sales=sales_data[:,2]
min_sales=np.min(all_sales)
print("Minimum sales in 2022: ",min_sales)

## Find the average sales for each restaurant across all 4 years.
all_year_sales = sales_data[:, 1:]
average_sales=np.mean(all_year_sales,axis=1)
print("\nAverage of sales of each restaurant across all 4 years : ",average_sales)

## Find the highest sales for each restaurant
all_year_sales = sales_data[:, 1:]
max_sales=np.max(all_year_sales,axis=1)
print("Highest sales of each restaurants:",max_sales)


##  Find the total sales across all restaurants for each year.
all_year_sales = sales_data[:, 1:]
total_sales=np.sum(all_year_sales,axis=0)
print("Total sales across all restaurant for each year: ",total_sales)



```

Sorting

```python
## Sort the restaurants by their 2024 sales.

sales_2024 = sales_data[:, -1]
sorted_restaurants = sales_data[np.argsort(sales_2024)]

print("Restaurants sorted by their 2024 sales:")
print(sorted_restaurants)


##Find the top 3 restaurants by 2024 sales.

sales_2024 = sales_data[:, -1]

sorted_restaurants = sales_data[np.argsort(sales_2024)]
top_3=sorted_restaurants[-3:,:]
print("Top 3 by 2024 sales: ",top_3)

##Use np.argsort() to get the ranking of restaurants by 2024 sales.

ranked_restaurants = sales_data[np.argsort(sales_2024), 0]

print("Restaurants ranked by their 2024 sales:")
print(ranked_restaurants)
```

Filtering with masks

```python
## Find restaurants whose 2024 sales are greater than 250000
sales_2024 = sales_data[:,-1]
mask=sales_2024 > 250000
filtered_restaurants=sales_data[mask,0] ## if asked to find sales greater than 250000 then use sales_2024[mask]
print("Filtered restaurants:",filtered_restaurants)



# Find restaurants whose 2024 AND 2023 sales are both greater than 200000.

sales_2024 = sales_data[:, -1]
sales_2023 = sales_data[:, -2]

mask = (sales_2024 > 200000) & (sales_2023 > 200000)

filtered_restaurants = sales_data[mask, 0]

print("Restaurants whose 2024 AND 2023 sales are greater than 200000:")
print(filtered_restaurants)


## Find restaurants whose 2024 sales are greater than 250000 OR 2021 sales are greater than 200000.
sales_2024 = sales_data[:, -1]
sales_2021=sales_data[:,1]

mask=(sales_2024 > 2500000) | (sales_2021 > 200000)

filtered_restaurants=sales_data[mask,0]

print("Restaurants whose 2024 sales are greater than 250k and 2021 is 200k",filtered_restaurants)


##  Get the restaurant IDs where 2024 sales are greater than 250000
sales_2024 = sales_data[:, -1]

mask=sales_2024 > 250000

filtered_restaurants=sales_data[mask,0]
print("Restaurants ID whose 2024 sales are greater than 250000:")
print(filtered_restaurants)
```
