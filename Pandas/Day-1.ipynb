Series

```python
import pandas as pd

#Creating series
sales=pd.Series([1,2,3,4,5])
print(sales)


##indexing in series
print(sales[2])

##Adding custom labels ans accessing the elements
sales=pd.Series([1,2,3,4],index=["a","b","c","d"])
print(sales["c"])

#Slicing
s = pd.Series([10, 20, 30, 40, 50])

print(s[1:4]) ##last index is not taken into account in slicing

##Methods and attributes in pandas

s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s)
print(s.index)    # Index(['a', 'b', 'c', 'd'])
print(s.values)   # [10 20 30 40]
print(s.dtype)    # int64 (typically)
print(s.shape)    # (4,)
print(s.size)     # 4
print(s.ndim)     # 1

print(s.head(2))   # First 2 elements
print(s.tail(2))   # Last 2 elements
print(s.describe())
print(s.to_list()) # [10, 20, 30, 40]

print(s.loc['a'])       # 10
print(s.iloc[2])        # 30
print(s.isin([10, 30])) # True, False, True, False
print(s.between(20, 40))
```

Dataframes

```python
import pandas as pd

data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Sleep_Hours': [8, 7, 7, 6, 6],
    'Attendance': [60, 70, 80, 90, 95],
    'Score': [40, 50, 65, 80, 95]
}
df = pd.DataFrame(data)

# ---------------- .loc (by label) ----------------
print(df.loc[2, 'Score'])                          # single value: row 2, Score
print(df.loc[1:3])                                 # rows 1 to 3 (end INCLUDED)
print(df.loc[:, ['Study_Hours', 'Score']])         # all rows, selected columns

# ---------------- .iloc (by position) ----------------
print(df.iloc[2, 3])                               # row position 2, column position 3
print(df.iloc[1:3])                                # rows 1 and 2 (end EXCLUDED)
print(df.iloc[:, :-1])                             # all columns except the last (features X)

# ---------------- Filtering ----------------
print(df[df['Score'] > 60])                                  # one condition
print(df[(df['Score'] > 50) & (df['Attendance'] > 80)])      # two conditions with & (each in brackets)
print(df.loc[df['Score'] > 60, ['Study_Hours', 'Score']])    # filter rows + pick columns # bottom 2 rows by Score      
                                                             
                                                             
                                                             
# ---------------- Adding ----------------
df['Result'] = ['Fail', 'Fail', 'Pass', 'Pass', 'Pass']      # new column
df['Study_Per_Sleep'] = df['Study_Hours'] / df['Sleep_Hours']  # new column from calculation
df.loc[5] = [12, 5, 98, 99, 'Pass', 2.4]                     # new row (label 5)
print(df)

# ---------------- Updating ----------------
df.loc[0, 'Score'] = 45                                      # update one value
df.loc[df['Score'] >= 65, 'Result'] = 'Pass'                 # update only rows matching a condition
df['Attendance'] = df['Attendance'] + 5                      # update a whole column
print(df)

# ---------------- Deleting ----------------
df = df.drop(columns=['Study_Per_Sleep'])                    # delete a column
df = df.drop(index=5)                                        # delete a row by label
df = df[df['Score'] >= 50]                                   # delete rows by condition (keep only Score >= 50)
print(df)                                                                        
```
